### Particle class
Onze `ParticleClass` definieert van een deeltje de massa $m$, snelheid $v$, positie $r$, en straal $R$. De positie moet per tijdstap bepaald worden. 

Nieuw aan `ParticleClass` zijn twee properties, nl. `momentum` en `energy`. Een property is eigenlijk een variabele die samenhangt met de waarde van andere variabelen binnen de klasse. Om de interne consistentie te bewaren moet je de juiste waarde van deze variabele dus telkens afleiden en dat doe je met een functie. Zo geeft `kin_energy` de kinetische energie terug van het deeltje door die af te leiden van de variabelen `m` en `v`.

Buiten `ParticleClass` voegen we twee functies toe, die je ook al hebt gezien. De eerste heet `collide_detection` en detecteert of twee deeltjes onderling overlappen en dus botsen. De tweede heet `particle_collision` en berekent de nieuwe snelheden van de twee deeltjes ten gevolge van hun botsing.

`````python
# Maken van de class
class ParticleClass:
    def __init__(self, m, v, r, R):
        """ maakt een deeltje (constructor) """
        self.m = m                         
        self.v = np.array(v, dtype=float)  
        self.r = np.array(r, dtype=float)  
        self.R = R

    def update_position(self):
        """ verandert positie voor één tijdstap """
        self.r += self.v * DT 
            
    @property
    def momentum(self):
        return self.m * self.v
    
    @property
    def kin_energy(self):
        return 1/2 * self.m * np.dot(self.v, self.v)
    
def collide_detection(p1: ParticleClass, p2: ParticleClass) -> bool:
    """ Geeft TRUE als de deeltjes overlappen """
    """ Geeft TRUE als de deeltjes overlappen """
    dx = p1.r[0]-p2.r[0]
    dy = p1.r[1]-p2.r[1]
    rr = p1.R + p2.R
    return  dx**2+dy**2 < rr**2 
    #Note: np.linalg.norm(p1.r - p2.r) < (p1.R + p2.R) had ook gekund, maar is veel langzamer. Controleer zelf!

def particle_collision(p1: ParticleClass, p2: ParticleClass):
    """ past snelheden aan uitgaande van overlap """
    m1, m2 = p1.m, p2.m
    delta_r = p1.r - p2.r
    delta_v = p1.v - p2.v
    dot_product = np.dot(delta_r, delta_v)
    
    # Als deeltjes van elkaar weg bewegen dan geen botsing
    if dot_product >= 0: # '='-teken voorkomt ook problemen als delta_r == \vec{0}
        return
        
    distance_squared = np.dot(delta_r, delta_r) 
    # Botsing oplossen volgens elastische botsing in 2D
    p1.v -= 2 * m2 / (m1 + m2) * dot_product / distance_squared * delta_r
    p2.v += 2 * m1 / (m1 + m2) * dot_product / distance_squared * delta_r
`````

### Volume en randvoorwaarden

We maken gebruik van een truc die in veel simulaties wordt gebruikt: we maken gebruik van periodieke randvoorwaarden. In dit geval wil dit zeggen dat een deeltje dat botst met de wand aan de ene kant van het volume, aan de andere kant verschijnt met exact dezelfde snelheid. Op deze manier verandert de snelheid van een deeltje niet bij het botsen met de wand en blijven alle behoudswetten gelden. Die behoudswetten kunnen we straks goed controleren.

`````python
def box_collision(particle: ParticleClass):
    ''' botsing met wanden geeft periodieke randvoorwaarden '''
    for i in range(2):
        if particle.r[i] > BOX_SIZE_0 / 2:
            particle.r[i] -= BOX_SIZE_0 
        elif particle.r[i] < -BOX_SIZE_0 / 2:
            particle.r[i] += BOX_SIZE_0
`````

### Maken van de simulatie

Nu dat we alle benodigde functies hebben, kunnen we ze in de juiste structuur zetten om de simulatie uit te voeren.

#### Bepaling positie deeltjes

Eerst moeten we alle deeltjes aanmaken en de startwaardes van hun variabelen invoeren:

```{solution} const_speed
Gelijke snelheid:
De code kiest eerst een willekeurige horizontale snelheid $v_x$ tussen $-V_0$ en $V_0$. Vervolgens wordt de verticale snelheid $v_y$ berekend met de stelling van Pythagoras: $v_y = \pm\sqrt{V_0^2 - v_x^2}$.
Hierdoor geldt voor de totale snelheid $v$:
$$v = \sqrt{v_x^2 + v_y^2} = \sqrt{v_x^2 + (V_0^2 - v_x^2)} = \sqrt{V_0^2} = V_0$$
Elk deeltje begint dus exact met de snelheid $V_0$.

Uniform verdeeld:
Voor de positie (`pos`) wordt de functie `np.random.uniform` gebruikt. Deze functie kiest willekeurige coördinaten ($x$ en $y$) waarbij elke plek binnen de opgegeven grenzen (de grootte van de doos minus de radius) een even grote kans heeft om gekozen te worden.
```

`````python
def create_particles(particles):
    ''' Leegmaken en opnieuw aanmaken van deeltjes '''
    particles.clear
    for i in range(N):
        vx = np.random.uniform(-V_0, V_0)
        vy = np.random.choice([-1, 1]) * np.sqrt(V_0**2 - vx**2)        
        pos = np.random.uniform(-BOX_SIZE_0/2 + RADIUS, BOX_SIZE_0/2 - RADIUS, 2)
        particles.append(ParticleClass(m=1.0, v=[vx, vy], r=pos, R=RADIUS))
`````

#### Controleren op onderlinge botsingen

Om de botsingen van alle deeltjes goed te verwerken moeten we controleren welke paren deeltjes met elkaar moeten botsen en deze botsingen één keer uitvoeren. 

```{exercise}
:label: double_loop
Leg uit hoe er in onderstaande code wordt voorkomen dat de botsing tussen twee deeltjes dubbel wordt uitgevoerd.
```


```{solution} double_loop
Dit wordt voorkomen door de grenzen van de loops (`range`).

De buitenste loop itereert met index `i` door alle deeltjes. De binnenste loop itereert met index `j`, maar begint pas bij `i + 1`.

Hierdoor wordt deeltje `i` alleen vergeleken met deeltjes die verderop in de lijst staan. Een paar `(A, B)` wordt dus wel gecontroleerd (wanneer `i` bij A is en `j` bij B), maar het omgekeerde paar `(B, A)` wordt overgeslagen (want wanneer `i` later bij B aankomt, kijkt `j` alleen naar deeltjes die ná B komen). Zo wordt elke unieke combinatie van twee deeltjes precies één keer behandeld.
```

`````python
def handle_collisions(particles):
    """ alle onderlinge botsingen afhandelen voor deeltjes in lijst """
    num_particles = len(particles)
    for i in range(num_particles):
        for j in range(i+1, num_particles):
            if collide_detection(particles[i], particles[j]):
                particle_collision(particles[i], particles[j])
`````

#### Controleren op botsingen met wanden

Voor het bepalen of deeltjes botsen met de wanden maken we gebruik van de functie die we net hebben aangepast.

`````python
def handle_walls(particles):
    ''' botsing met wanden controleren voor alle deeltjes in volume '''
    for p in particles:
        box_collision(p)
`````

#### Integreren in tijdstap

Nu moeten we alle functies op de juiste manier combineren om de simulatie een tijdstap te laten maken:

`````python
def take_time_step(particles):
    for p in particles:
        p.update_position()
        
    handle_collisions(particles)
    handle_walls(particles)  
`````

### Uitvoeren simulatie en tonen resultaat

Zoals aangegeven, kunnen we een animatie maken van de positie en snelheid als functie van de tijd, maar we kunnen ook het eindresultaat tonen en interpreteren:

`````python
particles = []
create_particles(particles)
for i in range(100):
    take_time_step(particles)

plt.figure()
plt.xlabel('x')
plt.ylabel('y')
plt.gca().set_aspect('equal')
plt.xlim(-BOX_SIZE_0/2, BOX_SIZE_0/2)
plt.ylim(-BOX_SIZE_0/2, BOX_SIZE_0/2)

for p in particles:
    plt.plot(p.r[0], p.r[1], 'k.', ms=25)
    plt.arrow(p.r[0], p.r[1], p.v[0], p.v[1], 
              head_width=0.01 * BOX_SIZE_0, head_length=0.015 * BOX_SIZE_0, color='red')
plt.show()
`````

```{solution} ex-recap-9
De buitenste lus (i) gaat elk deeltje af.
De binnenste lus (j) begint bij i+1. Je checkt daardoor een deeltje niet met zichzelf. Je checkt elk paar maar één keer (als je deeltje 1 met 5 hebt gecheckt, hoef je later niet deeltje 5 met 1 te checken). Zodra een botsing wordt gedetecteerd (if collide...), wordt deze meteen afgehandeld (particle_collision). De snelheden van die twee deeltjes veranderen dus direct, nog voordat de lus klaar is met de andere deeltjes.
```



`````python
import math

class Particle:
    def __init__(self, name, x, vx):
        self.name = name
        self.x = x
        self.vx = vx
        self.radius = 1.0

    def __repr__(self):
        return f"{self.name}: x={self.x:.2f}, v={self.vx:.2f}"

def collide_detection(p1, p2):
    # afstand < som van stralen
    dist = abs(p1.x - p2.x)
    return dist <= (p1.radius + p2.radius)

def particle_collision(p1, p2):
    print(f"  Botsing tussen {p1.name} en {p2.name}")
    # Elastische botsing (1D, gelijke massa) -> snelheden wisselen
    p1.vx, p2.vx = p2.vx, p1.vx
    print(f"  Nieuwe snelheden: {p1} | {p2}")

pA = Particle("A", x=-2.0, vx=10)
pC = Particle("C", x=0.0,  vx=0)
pB = Particle("B", x=2.0,  vx=-10)

particles = [pA, pC, pB] 

print("VOOR BOTSINGEN")
for p in particles: print(p)

print("\nLOOP START ")
num_particles = len(particles)
for i in range(num_particles):
    for j in range(i+1, num_particles):
        print(f"Check paar: {particles[i].name} & {particles[j].name}")
        
        if collide_detection(particles[i], particles[j]):
            particle_collision(particles[i], particles[j])

print("\nNA DE LOOP")
for p in particles: print(p)
`````

```{solution} ex-recap-9
Als je het bovenstaande script draait, zie je het volgende gebeuren:

De loop checkt eerst A en C. Die botsen. A staat nu stil (v=0), C krijgt de snelheid van A (v=10) en beweegt nu naar rechts.
De loop gaat verder en checkt uiteindelijk C en B. C heeft nu snelheid 10 (naar rechts). B heeft snelheid -10 (naar links). Ze botsen ook. C krijgt snelheid -10, B krijgt snelheid 10.
De uitkomst voor A: In een echte gelijktijdige 3-deeltjes botsing zou A misschien terugkaatsen. In deze code staat A echter stil na de interactie met C. A "weet niet" dat B er ook was.
Het systeem behandelt botsingen sequentieel (na elkaar), niet simultaan. De staat van deeltje C is al veranderd voordat deeltje B aan de beurt is. 
Er vinden slechts paarsgewijze interacties plaats. Een "tri-collision" bestaat niet in dit model; het wordt opgelost als twee losse botsingen vlak na elkaar.

In de computationele fysica is dit model om twee redenen acceptabel:
Kansberekening: De kans dat 3 deeltjes op exact hetzelfde tijdstip ($dt \to 0$) op exact hetzelfde punt botsen is infinitesimaal klein. In een systeem met veel ruimte tussen de deeltjes (een "verdund" systeem) zijn 3-deeltjes botsingen statistisch verwaarloosbaar.
Tijdsresolutie: Als je tijdstap ($dt$) klein genoeg is, worden botsingen die "bijna" tegelijk gebeuren, uitgesmeerd over verschillende tijdstappen. De fout die je maakt door ze sequentieel in dezelfde stap te behandelen is vaak kleiner dan de integratiefout van de bewegingsvergelijkingen zelf.
```

`````python
# Aangepaste functie die botsingen telt
def handle_collisions_with_counter(particles):
    botsingen_teller = 0
    num_particles = len(particles)
    for i in range(num_particles):
        for j in range(i+1, num_particles):
            if collide_detection(particles[i], particles[j]):
                particle_collision(particles[i], particles[j])
                botsingen_teller += 1
    return botsingen_teller

# Nieuwe simulatie runnen
particles = []
create_particles(particles) # Reset deeltjes
botsingen_per_stap = []

for i in range(200): # 200 tijdstappen
    take_time_step(particles)
    
    aantal = handle_collisions_with_counter(particles)
    botsingen_per_stap.append(aantal)

# 3. Plotten
plt.figure(figsize=(10, 4))
plt.bar(range(200), botsingen_per_stap, color='orange')
plt.xlabel('Tijdstap')
plt.ylabel('Aantal botsingen')
plt.title('Botsingsfrequentie in het gas/vloeistof')
plt.show()
`````
